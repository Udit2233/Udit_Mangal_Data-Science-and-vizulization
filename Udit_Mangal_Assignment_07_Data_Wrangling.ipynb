```python
res_first = marks_orig_idx['Marks'].combine_first(marks_upd_idx['Marks'])
res_fmax = marks_orig_idx['Marks'].combine(marks_upd_idx['Marks'], np.fmax)

comparison = pd.DataFrame({
    'Original': marks_orig_idx['Marks'],
    'Updated': marks_upd_idx['Marks'],
    'Combine_First': res_first,
    'Combine_Fmax': res_fmax
})
print(comparison)
differing_ids = comparison[
    (comparison['Combine_First'] != comparison['Combine_Fmax']) &
    ~(comparison['Combine_First'].isna() & comparison['Combine_Fmax'].isna())
].index.tolist()

print(differing_ids)
```
