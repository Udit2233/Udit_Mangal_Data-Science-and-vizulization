# **Section B**

Q16.

```python
df.groupby("branch")["units_sold"].agg(["sum", "mean", "min", "max", "std"])
```

Q17.

```python
df.groupby("product").agg(
    total_units=("units_sold", "sum"),
    avg_discount=("discount_pct", "mean"),
    max_rating=("customer_rating", "max"),
    min_delivery=("delivery_minutes", "min"),
)
```

Q18.

```python
df.groupby("branch").agg(
    total_records=("units_sold", "size"),
    non_missing_ratings=("customer_rating", "count")
)
```

Q19.

```python
def range_func(x):
    return x.max() - x.min()

df.groupby("product")["units_sold"].agg(range_func)
```

Q21.

```python
for branch_name, group_df in df.groupby('branch'):
    print(f"=== Branch: {branch_name} ===")
    print(group_df)
    print(f"Total Units Sold: {group_df['units_sold'].sum()}\n")
```

Q22.

```python
percentiles = df.groupby('product')['units_sold'].quantile([0.25, 0.50, 0.75]).unstack()
percentiles.columns = ['25th_Percentile', '50th_Percentile', '75th_Percentile']
percentiles
```

Q23.

```python
df_multi = df.set_index(['branch', 'product'])
branch_totals = df_multi.groupby(level='branch')['units_sold'].sum()
branch_totals
```

Q24.

```python
pivot_total_units = df.pivot_table(
    index='branch',
    columns='product',
    values='units_sold',
    aggfunc='sum',
    margins=True,
    margins_name='Total'
)
pivot_total_units
```

Q25.

```python
pivot_avg_metrics = df.pivot_table(
    index='branch',
    columns='product',
    values=['discount_pct', 'customer_rating'],
    aggfunc='mean'
)
pivot_avg_metrics
```

# **Section C**

Q26.

```python
print(df[df['customer_rating'].isna()])

df['customer_rating'] = df['customer_rating'].fillna(
    df.groupby(['branch', 'product'])['customer_rating'].transform('mean')
)

print(df)
print(df['customer_rating'].isna().sum())
```

Q27.

```python
def weighted_avg_discount(group):
    return (group['units_sold'] * group['discount_pct']).sum() / group['units_sold'].sum()

weighted_discounts = df.groupby('branch').apply(weighted_avg_discount, include_groups=False)
weighted_discounts
```

Q28.

```python
top_3_sales = (
    df.groupby('branch', group_keys=False)
      .apply(lambda g: g.nlargest(3, 'units_sold'))
      [['branch', 'product', 'day', 'units_sold']]
)
top_3_sales
```

Q29.

```python
branch_total = df.groupby('branch')['units_sold'].transform('sum')
df['sales_contribution_pct'] = (df['units_sold'] / branch_total) * 100

print(df[['record_id', 'branch', 'units_sold', 'sales_contribution_pct']].head(10))
print(df.groupby('branch')['sales_contribution_pct'].sum())
```

Q30.

```python
labels = ['Low Sales', 'Below Average Sales', 'Above Average Sales', 'High Sales']
df['sales_category'] = pd.qcut(df['units_sold'], q=4, labels=labels)

category_summary = df.groupby('sales_category', observed=False).agg(
    total_units_sold=('units_sold', 'sum'),
    avg_discount_pct=('discount_pct', 'mean'),
    avg_customer_rating=('customer_rating', 'mean'),
    avg_delivery_minutes=('delivery_minutes', 'mean'),
    sales_records_count=('record_id', 'count')
)
category_summary
```
